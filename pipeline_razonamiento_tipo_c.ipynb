# Pipeline de razonamiento y espacio de estados — Tipo C
## Caso: patrones sintéticos de sueño en recién nacidos

**Propósito:** representar el flujo lógico de un proyecto de análisis de datos. No es una herramienta clínica. Todos los datos utilizados en este proyecto son sintéticos.

## 1. Definición del problema
Se desea demostrar cómo un pipeline de aprendizaje automático puede procesar datos ficticios de sueño durante un periodo de 24 horas y estimar una variable numérica (`sueno_total_h`). El objetivo es educativo: practicar modelado, validación y visualización, no evaluar la salud de un recién nacido.

```python
# Espacio de estados, operadores y transiciones

estados = {
    "S0": "Inicio",
    "S1": "Datos sintéticos generados",
    "S2": "Datos validados",
    "S3": "Datos explorados y preparados",
    "S4": "Modelo de regresión entrenado",
    "S5": "Predicciones evaluadas y visualizadas",
}

operadores = {
    "O1": "Generar datos sintéticos",
    "O2": "Validar rangos y consistencia",
    "O3": "Explorar y preparar variables",
    "O4": "Entrenar modelo de regresión",
    "O5": "Evaluar y visualizar resultados",
}

transiciones = [
    ("S0", "O1", "S1"),
    ("S1", "O2", "S2"),
    ("S2", "O3", "S3"),
    ("S3", "O4", "S4"),
    ("S4", "O5", "S5"),
]

print("ESTADOS")
for codigo, descripcion in estados.items():
    print(f"{codigo}: {descripcion}")

print("\nOPERADORES")
for codigo, descripcion in operadores.items():
    print(f"{codigo}: {descripcion}")

print("\nTRANSICIONES")
for origen, operador, destino in transiciones:
    print(f"{origen} --{operador}--> {destino}")
```

## 2. Restricciones
1. `edad_dias` debe estar entre 0 y 28, porque el escenario sintético se limita a los primeros 28 días.
2. Los conteos deben ser números no negativos.
3. Las horas de sueño diurno y nocturno deben estar entre 0 y 24.
4. La suma de sueño diurno y nocturno debe ser igual a `sueno_total_h`.
5. No se deben interpretar las variables ni las predicciones como criterios médicos.

```python
# Recorrido de búsqueda desde el estado inicial hasta el objetivo

estado_actual = "S0"
estado_objetivo = "S5"
ruta = [estado_actual]

for origen, operador, destino in transiciones:
    if origen == estado_actual:
        print(f"Aplicando {operador}: {estados[origen]} -> {estados[destino]}")
        estado_actual = destino
        ruta.append(estado_actual)

print("\nRuta recorrida:", " -> ".join(ruta))
print("Objetivo alcanzado:", estado_actual == estado_objetivo)
```

## 3. Diagrama textual del pipeline

`S0 Inicio → S1 Generación → S2 Validación → S3 Exploración/preparación → S4 Entrenamiento → S5 Evaluación/visualización`

## 4. Resultado esperado
El notebook muestra los estados, operadores, restricciones y la ruta secuencial hasta el estado objetivo. Este recorrido formaliza el proceso de trabajo; no implica que se haya descubierto un resultado clínico.
